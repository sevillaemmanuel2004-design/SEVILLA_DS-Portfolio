**Instruction:** Do one forward pass through the network and compute the error.

Input $x = [1, 0, 1]^T$, target $y = 1$, activation ReLU: $f(z) = \max(0, z)$.

## Given values

Weights from the inputs to the hidden layer: $w_{11}=0.2,\ w_{12}=-0.3,\ w_{13}=0.4,\ w_{14}=0.1,\ w_{15}=-0.5,\ w_{16}=0.2$


Weights from the hidden layer to the output: $w_{21}=-0.3,\ w_{22}=-0.2$

Biases: $\theta_1=-0.4,\ \theta_2=0.2,\ \theta_3=0.1$

## Step 1 — Hidden layer

Hidden neuron 1:
$Z_1 = (0.2)(1) + (0.4)(0) + (-0.5)(1) + (-0.4) = -0.7$
$H_1 = \max(0, -0.7) = 0$

Hidden neuron 2:
$Z_2 = (-0.3)(1) + (0.1)(0) + (0.2)(1) + 0.2 = 0.1$
$H_2 = \max(0, 0.1) = 0.1$

## Step 2 — Output layer

$Z_3 = (-0.3)(0) + (-0.2)(0.1) + 0.1 = 0.08$
$\hat{y} = \max(0, 0.08) = 0.08$

## Step 3 — Error

Following the lesson, the error is $\delta = \hat{y} - y$:

$\delta = 0.08 - 1 = \mathbf{-0.92}$

For reference, the squared error is $(1 - 0.08)^2 = 0.8464$ and the half squared error is $0.4232$.

## Python check

```python
import numpy as np

x = np.array([1, 0, 1])
y = 1

# columns = hidden neurons, rows = inputs
W1 = np.array([[0.2, -0.3],
               [0.4,  0.1],
               [-0.5, 0.2]])
b1 = np.array([-0.4, 0.2])
W2 = np.array([-0.3, -0.2])
b2 = 0.1

def relu(z):
    return np.maximum(0, z)

z_hidden = x @ W1 + b1
h = relu(z_hidden)
z_out = h @ W2 + b2
y_hat = relu(z_out)
error = y_hat - y

print("Hidden Z :", z_hidden)
print("Hidden H :", h)
print("Output Z :", round(z_out, 4))
print("y_hat    :", round(y_hat, 4))
print("Error    :", round(error, 4))
print("Squared  :", round(error ** 2, 4))
```
