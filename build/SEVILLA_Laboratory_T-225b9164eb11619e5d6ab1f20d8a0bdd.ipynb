## 1. Standard imports

```python
import numpy as np
import torch

print("NumPy  :", np.__version__)
print("PyTorch:", torch.__version__)
```

## 2. Create `set_seed(seed: int)`

The function sets the random seed for NumPy and PyTorch. It returns nothing.

```python
def set_seed(seed: int) -> None:
    np.random.seed(seed)
    torch.manual_seed(seed)
```

## 3. Create the NumPy array `arr`

Six random integers from 0 (inclusive) to 5 (exclusive), using seed 42. The dtype is `int32` so Step 5 can change it to `int64`.

```python
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype("int32")

print("arr   :", arr)
print("dtype :", arr.dtype)
```

## 4. Make tensor `x` from `arr`

```python
x = torch.from_numpy(arr)

print(x)
print("dtype:", x.dtype)
```

## 5. Change `x` from `int32` to `int64`

```python
x = x.long()

print(x)
print("dtype:", x.dtype)
```

## 6. Reshape `x` to 3 × 2

```python
x = x.reshape(3, 2)

print(x)
print("shape:", x.shape)
```

## 7. Get the right-hand column of `x`

```python
right_col = x[:, 1]
print(right_col)
```

## 8. Square the values of `x` without changing `x`

`x ** 2` makes a new tensor, so `x` stays the same.

```python
squared = x ** 2

print("Squared:")
print(squared)
print("\nx is still:")
print(x)
```

## 9. Create tensor `y`

`x` is 3 × 2, so `y` needs 2 rows to be multiplied with `x`. To keep 6 elements like `x`, `y` is 2 × 3.

```python
torch.manual_seed(42)
y = torch.randint(0, 5, (2, 3))

print(y)
print("shape   :", y.shape)
print("elements:", y.numel())
```

## 10. Matrix product of `x` and `y`

A 3 × 2 matrix times a 2 × 3 matrix gives a 3 × 3 matrix.

```python
product = x @ y

print(product)
print("shape:", product.shape)
```

## Summary

All 10 steps are done: imports, `set_seed()`, the seeded NumPy array, conversion to a tensor, `int64` change, 3 × 2 reshape, right column, squared values, a 2 × 3 tensor `y`, and the matrix product.
