## Setup

```python
import numpy as np

x = np.array([1, 0, 1])
y = 1
lr = 0.001

W1 = np.array([[0.2, -0.3],
               [0.4,  0.1],
               [-0.5, 0.2]])
b1 = np.array([-0.4, 0.2])
W2 = np.array([-0.3, -0.2])
b2 = 0.1

def relu(z):
    return np.maximum(0, z)

def relu_grad(z):
    return (z > 0) * 1.0
```

## 1. Forward propagation

```python
z1 = x @ W1 + b1
h = relu(z1)
z2 = h @ W2 + b2
y_hat = relu(z2)

error = y_hat - y
loss = 0.5 * error ** 2

print("z1    :", z1)
print("h     :", h)
print("z2    :", round(z2, 4))
print("y_hat :", round(y_hat, 4))
print("error :", round(error, 4))
print("loss  :", round(loss, 4))
```

## 2. Backward propagation

With the loss $E = \frac{1}{2}(\hat{y} - y)^2$, the output delta is $(\hat{y} - y)\,f'(Z_3)$.
The delta is then sent back to the hidden layer through the output weights and the ReLU derivative.

```python
# output layer
delta_out = error * relu_grad(z2)
dW2 = h * delta_out
db2 = delta_out

# hidden layer
delta_hidden = W2 * delta_out * relu_grad(z1)
dW1 = np.outer(x, delta_hidden)
db1 = delta_hidden

print("delta_out    :", round(delta_out, 4))
print("delta_hidden :", delta_hidden)
print("dW2:", dW2)
print("db2:", round(db2, 4))
print("dW1:")
print(dW1)
print("db1:", db1)
```

## 3. Update the weights and biases

```python
W1_new = W1 - lr * dW1
b1_new = b1 - lr * db1
W2_new = W2 - lr * dW2
b2_new = b2 - lr * db2

print("New W1:")
print(W1_new)
print("New b1:", b1_new)
print("New W2:", W2_new)
print("New b2:", round(b2_new, 6))
```

## 4. Check after one update

```python
z1 = x @ W1_new + b1_new
h = relu(z1)
z2 = h @ W2_new + b2_new
y_hat_new = relu(z2)
loss_new = 0.5 * (y_hat_new - y) ** 2

print("New prediction:", round(y_hat_new, 5))
print("New loss      :", round(loss_new, 6))
print("Old loss      :", round(loss, 6))
print("Loss decreased:", loss_new < loss)
```
